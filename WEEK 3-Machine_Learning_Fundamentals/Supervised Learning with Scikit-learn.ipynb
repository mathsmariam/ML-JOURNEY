# Introduction to Machine Learning

Machine learning is the process whereby computers learn to make decisions from data, without being explicitly programmed for every situation. Instead of writing fixed rules to handle each possible case, you give the computer examples, and it learns the underlying patterns on its own, then uses those patterns to make predictions or decisions on new data it hasn't seen before.

This is a shift from traditional programming, where a human writes exact instructions for every outcome. In machine learning, the "instructions" are learned from the data itself.

There are three main types of machine learning:

Supervised Learning:  learning from labeled data (you already know the correct answers), used to predict outcomes for new data. Example: predicting house prices based on past sales.

Unsupervised Learning: learning from unlabeled data, finding hidden patterns or groupings on its own. Example: grouping customers with similar buying habits.

Reinforcement Learning: learning through trial and error, receiving rewards or penalties for actions taken. Example: an AI learning to play a game by improving with each attempt.

# Machine Learning Algorithm

### Supervised Learning Algorithms

1. Linear Regression: A statistical method used to model the relationship between a dependent variable(target) and one or more  independent variables(predictors). Applications: Real Estate Price Prediction, Sales Forecasting, Stock Prices Prediction.

2. Logistic Regression: A classification algorithm used to predict binary outcome (e,g YES/NO, TRUE/FALSE). It uses a logistic function to model the probability of a particular class. Applications: Email Spam Filtering, Medical Diagnosis, Customer Churn Prediction.

3. Decision Trees: A flowchart-like tree structure used to make decisions. Each node in th tree represents a decision based on a feature, and each branch represent a possible outcome. Applications: Loan Approval, Medical Diagnosis, Marketng Campaign Analysis.

4. Random Trees: An ensemble method that combines multiple decision trees to improve accuracy. Each tree gives a vote on the outcome, and the majority vote determines the final decision. Applications: Health Care Disease Prediction, Fraud Detection, Customer segmentation.

5. Support Vector Machine (SVM): A classification algorithm that finds the optimal boundary(hyperplane) to seperate data into different classes, often used for binary classification. Applications: Facial Recognition, Speech Recognition, Handwritten Digit Recognition.

### Unsupervised Learning Algorithm

1. K-Means Clustering: A clustering algorithm that groups data into a specified number ($K$) of clusters based on similarity. The goal is to minimize the distance between data points in each cluster. Applications: Customer segmentation in marketing, Market basket analysis, Social media grouping.

2. Hierarchical Clustering: A clustering algorithm that creates a tree-like structure (dendrogram) by grouping similar data points iteratively, which can be either agglomerative (bottom-up) or divisive (top-down). Applications: Gene segmentation, Gene expression analysis, Customer behaviour analysis, Document clustering.

3. DBSCAN (Density-Based Spatial Clustering of Applications with Noise): A density-based clustering algorithm that identifies clusters based on the density of data points. It can also handle noise (outliers) by labeling them as noise points. Applications: Crime hotspot detection, Geospatial analysis, Anomaly detection.

4. Principal Component Analysis (PCA): A dimensionality reduction technique that transforms data into a smaller set of uncorrelated variables (principal components) to capture the most variance in the data. Eg Image Compression. Applications: Data compression, Dimensionality reduction, Data visualization.

5. Autoencoders: A type of neural network used to learn efficient representations of data, typically for dimensionality reduction or anomaly detection. Applications: Fraud detection, Image denoising, Recommendation systems.

# K-Nearest Neighbors

What is k-NN?

K-Nearest Neighbor (k-NN) is a simple yet widely-used classification algorithm. It can also be used for regression.

k-NN is:

Non-parametric: it doesn't make any assumptions about the underlying shape or distribution of the data (unlike something like linear regression, which assumes a linear relationship)

Instance-based: it doesn't explicitly learn a model or formula. Instead, it simply memorizes the training data itself

A supervised learning algorithm: it needs labeled data to work

Because it doesn't build a model upfront and instead just stores the data, k-NN is also called a lazy algorithm because it is instance based, all the real computation happens later, at prediction time, rather than during training.

k-NN is a way for a computer to guess something new by looking at things it already knows that are most similar to it.

It doesn't try to learn a formula or rule. It just remembers every example it's seen before. When you give it something new, it looks at the closest examples to that new thing, and copies the answer that most of them agree on.

That's why it's called "lazy", it doesn't do any real work until the moment you ask it to predict

```python
import pandas as pd
churn_df = pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')   # to load dataset
churn_df.head()
```

```python
# Check dataset size and full list of column names
print(churn_df.shape)          # (rows, columns)
print(churn_df.columns.tolist())   # list of every column name
```

```python
# Check for missing values
print(churn_df.isnull().sum())

# Check data types
print(churn_df.dtypes)
```

```python
# Find rows where TotalCharges is not a valid number

# Looking at the problem rows BEFORE fixing anything, pd.to_numeric() tries converting TotalCharges to a number.
# Any value that fails to convert becomes NaN (missing) - we use .isnull() to find exactly which rows those are, without permanently changing the column yet.

problem_rows = churn_df[pd.to_numeric(churn_df['TotalCharges'], errors='coerce').isnull()]
print(problem_rows[['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges']])
```

```python
# Convert TotalCharges to numeric, forcing bad value NaN for any non-numeric values. 
# This will allow us to handle those rows appropriately later (e.g., by dropping them or imputing values).
churn_df['TotalCharges'] = pd.to_numeric(churn_df['TotalCharges'], errors='coerce')

# Now check again for missing values or blank values in TotalCharges
print(churn_df['TotalCharges'].isnull().sum())
```

```python
churn_df = churn_df.dropna(subset=['TotalCharges']) # Drop rows where TotalCharges is NaN (missing) after conversion to numeric

print(churn_df['TotalCharges'].isnull().sum()) # Check again for missing values in TotalCharges after dropping rows with NaN
```

```python
print(churn_df.shape) # check the shape of the dataset after dropping rows with missing TotalCharges
```

```python
# Check for missing values again after dropping rows with NaN in TotalCharges
print(churn_df.isnull().sum())

# Check data types again after conversion
print(churn_df.dtypes)
```

### K-Nearest Neighbors: Fit and Predict

```python
# ===== Fit exercise =====
from sklearn.neighbors import KNeighborsClassifier

y = churn_df["Churn"].values   # capital C, matches your real dataset
X = churn_df[["tenure", "MonthlyCharges"]].values   # your actual numeric columns

# Create a KNN classifier with 6 neighbors
knn = KNeighborsClassifier(n_neighbors=6)

# Fit the classifier to the data (this is where it "memorizes" the training data)
knn.fit(X, y)
```

```python
import numpy as np

X_new = np.array([[12, 45.0],
                   [60, 90.0],
                   [3, 20.0]])

print(X_new.shape)
```

```python
# Predict churn for these new customers
y_pred = knn.predict(X_new)
print("Predictions: {}".format(y_pred))
```

### Measuring Model Performance

Measuring model performance

* In classification, accuracy is a commonly used metric

* Accuracy = correct predictions / total observations

How do we measure accuracy?

* We could compute accuracy on the data used to fit the classifier

* But this is NOT indicative of ability to generalize — it would tell you how well the model memorized data it already saw, not how well it performs on new data

### Train/test split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=21, stratify=y) # s

# Split data into training (70%) and test (30%) sets, random_state=21 makes the split reproducible (same split every run), 
# stratify=y ensures both sets have the same proportion of each class(e.g. same % of "Yes"/"No" churn in train and test, not skewed by chance)

# Create a KNN classifier with 6 neighbors
knn = KNeighborsClassifier(n_neighbors=6)

# Train the model using only the training set
knn.fit(X_train, y_train)

print(knn.score(X_test, y_test))  
# knn.score() does prediction AND accuracy calculation in one step: it predicts labels for X_test, compares them to the real y_test,
# and returns the accuracy (correct predictions / total observations)

```

```python
# Dictionaries to store accuracy results for each K value tested
train_accuracies = {}
test_accuracies = {}

# Try K values from 1 to 25
neighbors = np.arange(1, 26)

for neighbor in neighbors:
    # Create and train a KNN model with this specific K
    knn = KNeighborsClassifier(n_neighbors=neighbor)
    knn.fit(X_train, y_train)

    # Record accuracy on the TRAINING data (data it already learned from)
    train_accuracies[neighbor] = knn.score(X_train, y_train)

    # Record accuracy on the TEST data (unseen data - the real test)
    test_accuracies[neighbor] = knn.score(X_test, y_test)
   
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))   # set the plot size
plt.title("KNN: Varying Number of Neighbors")

# Plot training accuracy across all tested K values
plt.plot(neighbors, train_accuracies.values(), label="Training Accuracy")

# Plot test accuracy across all tested K values
plt.plot(neighbors, test_accuracies.values(), label="Testing Accuracy")

plt.legend()   # shows which line is which (using the labels above)
plt.xlabel("Number of Neighbors")
plt.ylabel("Accuracy")
plt.show()
```

```python
# Try a better K value based on what the chart showed
knn_best = KNeighborsClassifier(n_neighbors=12)   # or whichever K peaked for you
knn_best.fit(X_train, y_train)
print(knn_best.score(X_test, y_test))
```
