```python
# Exercise 4A
#  - write a program that ask for your grade
#  - prints out fail if your grade is lesser than 40
#  - print out pass if its higher thanS 40 but less than 50
#  - out credit if its higher than 50 but less than 70
#  - print out excellent if its higher than 70
# SOLUTION TO 4A
grade = int(input("Enter your grade: "))
if grade > 0 and grade < 40:
  print("fail")
elif grade >= 40 and grade < 50:
    print("pass")
elif grade >= 50 and grade < 70:
      print("credit")
elif grade >= 70 and grade <= 100:
  print("excellent")
else:
    print( 'enter number between 1 to 100' )
```

```python
# EXERCISE 4B
# DESIGN THE UNILAG GRADING SYSTEM USING PYTHON
# YOUR PRINT OUT STATEMENT SHOULD LOOK LIKE THIS
# YOU HAD 21 IN YOUR CA AND 50 IN YOUR EXAM. THATS A GRADE OF A
# SOLUTION TO 4B
 
CA = int(input("Enter your CA score: "))
EXAM = int(input("Enter your Exam score: "))

if not (0 <= CA <= 30):
    print("Invalid CA score! CA must be between 0 and 30.")
elif not (0 <= EXAM <= 70):
    print("Invalid Exam score! Exam must be between 0 and 70.")
else:
    TOTAL = CA + EXAM
    if TOTAL >= 70:
        grade = 'A'
    elif TOTAL >= 60:
        grade = 'B'
    elif TOTAL >= 50:
        grade = 'C'
    elif TOTAL >= 45:
        grade = 'D'
    elif TOTAL >= 40:
        grade = 'E'
    else:
        grade = 'F'

    print(f"You had {CA} in your CA and {EXAM} in your exam. That's a grade of {grade}.")
```
