#The if, else, elif statements

In Python, if, else, and elif (short for "else if") are used for conditional execution, allowing you to control the flow of your program based on certain conditions. Here's how they're used:

1. if statement: The if statement is used to execute a block of code only if a specified condition is True.

```python
A = 5
if A > 3:
    print("Yes")
```

2. else statement: The else statement is used to execute a block of code if the if condition is False.

```python
A = 5
if A > 2:
    print("Yes")
else:
    print('no')
```

3. elif statement(short for "else if"): The elif statement is used to check additional conditions if the preceding if or else conditions are False.

```python
A = 6
if A > 80:
    print("Yes")
elif A == 6:
    print("Good")
elif A % 5 == 0:
    print("Multiple of 5")
else:
    print('no')
```
